```python
import os
import requests
from bs4 import BeautifulSoup

# Base URL for the CSV files
base_url = "https://www.ncei.noaa.gov/pub/data/swdi/stormevents/csvfiles/"
headers = {'User-Agent': 'Mozilla/5.0'}

# Create a directory to save downloads
download_dir = "downloaded_csvs"
os.makedirs(download_dir, exist_ok=True)

# Request the directory listing
response = requests.get(base_url, headers=headers)
if response.status_code != 200:
    raise Exception(f"Failed to access {base_url} (status code: {response.status_code})")

soup = BeautifulSoup(response.text, 'html.parser')

csv_links = []
# Filter for links that end with .csv or .csv.gz (case-insensitive)
for link in soup.find_all('a'):
    href = link.get('href')
    if href and href.lower().endswith(('.csv', '.csv.gz')):
        csv_links.append(href)

print(f"Found {len(csv_links)} CSV files.")

# Download each file
for csv_file in csv_links:
    file_url = base_url + csv_file
    print(f"Downloading {file_url} ...")
    file_response = requests.get(file_url, headers=headers)
    if file_response.status_code == 200:
        file_path = os.path.join(download_dir, csv_file)
        with open(file_path, 'wb') as f:
            f.write(file_response.content)
    else:
        print(f"Failed to download {file_url} (status code: {file_response.status_code})")

print("Download complete.")
```

```python
import os
import pandas as pd

# Directory where the CSV (or CSV.GZ) files are stored
download_dir = "/Users/heonkim/Desktop/data/downloaded_csvs"

# Get all CSV or CSV.GZ files in the directory
files = [f for f in os.listdir(download_dir) if f.lower().endswith(('.csv', '.csv.gz'))]

if not files:
    print("No CSV files found in the directory.")
else:
    for file in files:
        file_path = os.path.join(download_dir, file)
        print(f"\nReviewing file: {file}")
        try:
            # Read just the first few rows to inspect the columns
            df = pd.read_csv(file_path, compression='infer', nrows=10)
            
            # Display column names and data types
            print("Columns and their data types:")
            for col, dtype in zip(df.columns, df.dtypes):
                print(f" - {col}: {dtype}")
            
            # Optionally, print the first few rows to get a sense of the data
            print("\nPreview of the data:")
            print(df.head())
            print("-" * 50)
            
        except Exception as e:
            print(f"Error reading {file}: {e}")
```

# Zillow files

```python
import pandas as pd

def download_zillow_csv(url):
    """
    Download a CSV file from the provided Zillow URL, print its columns and a preview,
    and return the DataFrame.
    """
    try:
        df = pd.read_csv(url)
        print(f"Data loaded from: {url}")
        print("Columns:", list(df.columns))
        print("Preview:")
        print(df.head(), "\n")
        return df
    except Exception as e:
        print(f"Error loading data from {url}: {e}")
        return pd.DataFrame()

# Zillow CSV URLs
urls = [
    "https://files.zillowstatic.com/research/public_csvs/zhvi/Metro_zhvi_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.csv?t=1739388659",
    "https://files.zillowstatic.com/research/public_csvs/zori/Metro_zori_uc_sfrcondomfr_sm_month.csv?t=1739388659",
    "https://files.zillowstatic.com/research/public_csvs/mean_doz_pending/Metro_mean_doz_pending_uc_sfrcondo_sm_month.csv?t=1739388659"
]

# Download and store each DataFrame in a dictionary
zillow_dfs = {}
for url in urls:
    # Extract a filename from the URL (removing the query parameters)
    key = url.split('/')[-1].split('?')[0]  
    zillow_dfs[key] = download_zillow_csv(url)

# Save each DataFrame to a CSV file in the current working directory
for key, df in zillow_dfs.items():
    output_filename = key  # Filename based on the URL-derived key
    df.to_csv(output_filename, index=False)
    print(f"Saved {output_filename} to disk.")
```
